## Neural Network Fits

```python
datafed_path = "2024_SHO_Fitting/9_30_2024_Hysteresis_Fitting"

data, voltage = BE_viz.get_hysteresis(scaled=True, loop_interpolated=True)

X_train, X_test = train_test_split(
    data.reshape(-1, 96), test_size=0.2, random_state=42, shuffle=True
)

X_train = np.atleast_3d(X_train)

model_ = Multiscale1DFitter(
    hysteresis_nn,  # function
    voltage[:, 0].squeeze(),  # x data
    1,  # input channels
    9,  # output parameters
    BE_viz.loop_param_scaler,
    loops_scaler=BE_viz.hysteresis_scaler,
)

# instantiate the model
model = Model(
    model_,
    BE_viz,
    path="Trained Models/Hysteresis Fitter/",
    training=True,
    model_basename="Hysteresis_Loop_Fitter",
    datafed_path=datafed_path,
    script_path="./7_Hysteresis_Fitter.ipynb",
)

optimizer = {
    "name": "TRCG",
    "optimizer": TRCG,
    "closure_size": 1,
    "cgopttol": 1e-3,
    "c0tr": 0.2,
    "c1tr": 0.25,
    "c2tr": 0.75,
    "t1tr": 0.75,
    "t2tr": 2.0,
    "radius_max": 5.0,
    "radius_initial": 1.0,
    "radius": 1.0,
    "device": "cuda",
    "ADAM_epochs": 500,
}

#train = True
train = False

if train:
    # fits the model
    model.fit(
        X_train,
        1024,
        optimizer=optimizer,
        epochs=600,
    )
else:
    model.load(
        "/home/ferroelectric/Documents/m3_learning/m3_learning/papers/2023_Rapid_Fitting/Trained Models/Hysteresis Fitter/Hysteresis_Loop_Fitter_model_optimizer_Trust Region CG_epoch_599_train_loss_0.006924134410089917.pth"
    )
```
